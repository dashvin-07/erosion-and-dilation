# Morphological Operators

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def load_img():
    blank_img = np.zeros((300, 700), dtype=np.uint8)
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img,text='DASHVIN',org=(22,160),fontFace=font,fontScale= 5,color=(255,255,255),thickness=25,lineType=cv2.LINE_AA)
    return blank_img
```

```python
def display_img(img):
    fig = plt.figure(figsize=(8.4, 3))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap="gray")
    plt.show()
```

```python
img = load_img()
```

```python
display_img(img)
```

# Erosion

```python
img = load_img()
kernel = np.ones((5,5),np.uint8)
erosion5 = cv2.erode( img,kernel,iterations = 4)
```

```python
display_img(erosion5)
```

# Dilation

```python
img = load_img()
kernel = np.ones((5,5),np.uint8)
dilation5 = cv2.dilate(img, kernel, iterations = 4)
```

```python
display_img(dilation5)
```
